# **Ablation**

Os seguintes ablations servem para estudar e concluir sobre os impactes que as diferentes alterações na estratégia de treino, agregação, e decisão têm nos resulatdos obtidos no treino e validação da CNN2D.

------------------------------------------------------------------------------------------------------------------------------------------

**Pesos não suavizados e análise do:**
- impacte de na agregação passarmos a média simples das probabilidades de sopro presente em todas as janelas ou a média das 5 janelas com maior probabilidade de sopro presente
- impacte da decisão ser ponderada ou não, isto é, de multiplicarmos as pro0babilidades pelo vetor [5:3:1] antes de passar pelo argmax ou de passar apenas as probabilidades pelo argmax.
